```python
# Cell: Setup Project Root
import os
import sys
from pathlib import Path


def find_project_root(start_path=None):
    path = Path(start_path or os.getcwd()).resolve()
    for parent in [path, *path.parents]:
        if (parent / "src").is_dir() and (parent / "config").is_dir():
            return parent
    raise FileNotFoundError("Khong tim thay project root co thu muc src/ va config/")


project_root = find_project_root()
os.chdir(project_root)
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

print("Project root:", project_root)
```

```python
# Cell: Import Validation Modules
from src.data.loaders import read_symbol_files, save_parquet
from src.data.validators import standardize_ohlcv, validate_ohlcv
```

```python
# Cell: Clean And Validate Data
raw = read_symbol_files("data/raw/daily_ohlcv")
clean = standardize_ohlcv(raw)
checks = validate_ohlcv(clean)
print(checks)
clean.head()
```

```python
# Cell: Save Clean Dataset
save_parquet(clean, "data/processed/clean_daily_ohlcv.parquet")
clean.shape
```
