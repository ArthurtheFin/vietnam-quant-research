Notebook này là bản đồ vận hành của toàn bộ project theo flow mới:

1. Chuẩn bị môi trường và dữ liệu nền
2. Quét nhẹ toàn thị trường để chọn `auto top 10 watchlist`

5. Xuất dashboard và report cho khách hàng


```python
# Cell: Setup Project Root
import os
import sys
from pathlib import Path


def find_project_root(start_path=None):
    path = Path(start_path or os.getcwd()).resolve()
    for parent in [path, *path.parents]:
        if (parent / "src").is_dir() and (parent / "config").is_dir():
            return parent
    raise FileNotFoundError("Khong tim thay project root co thu muc src/ va config/")


project_root = find_project_root()
os.chdir(project_root)
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

print("Project root:", project_root)
```

## Notebook Responsibilities

### 01_validate_environment.ipynb
- Nhiệm vụ: kiểm tra môi trường chạy và sự sẵn sàng của data source.
- Đầu ra: trạng thái package, path, file, biến môi trường.

### 02_fetch_daily_ohlcv.ipynb
- Nhiệm vụ: fetch dữ liệu OHLCV, benchmark/index, securities universe.
- Đầu ra chính: Dữ liệu giá thô lưu trong data/raw/.

### 03_clean_validate_data.ipynb
- Nhiệm vụ: chuẩn hóa và làm sạch dữ liệu OHLCV thô.
- Đầu ra: clean_daily_ohlcv.parquet.

### 04_market_distribution.ipynb
- Nhiệm vụ: tính toán thống kê phân phối và tail ratio của giá/returns.
- Đầu ra: Các chỉ số thống kê phân phối lăn (rolling distribution stats).

### 05_volatility_regime.ipynb
- Nhiệm vụ: phân tích độ biến động (realized, EWMA) và phân loại volatility regime.
- Đầu ra: Phân loại trạng thái biến động thị trường.

### 06_market_regime.ipynb
- Nhiệm vụ: xác định xu hướng và các mức phân loại Market Regime (Bullish, Bearish, Neutral).
- Đầu ra: market_features.parquet chứa nhãn trạng thái thị trường.

### 07_breakout_research.ipynb
- Nhiệm vụ: đo hiệu quả của breakout setup cơ bản theo các khung nắm giữ khác nhau.
- Đầu ra: baseline trade performance summary.

### 07b_breakout_stoploss_regime.ipynb
- Nhiệm vụ: tối ưu hóa mức cắt lỗ fixed stop và ATR stop theo từng trạng thái biến động.
- Đầu ra: best stop candidates và volatility rule parameters.

### 08_monte_carlo_simulation.ipynb
- Nhiệm vụ: mô phỏng đường đi giá Monte Carlo để kiểm định xác suất chạm target trước khi chạm stop-loss.
- Đầu ra: fanchart, terminal distribution, exit probabilities.

### 09_stealth_accumulation_screen.ipynb
- Nhiệm vụ: tính toán điểm tích lũy âm thầm và phân loại giai đoạn (Stage S, Stage A, Stage B).
- Đầu ra: Bảng xếp hạng mã tích lũy và watchlist hành động.

### 10_benchmark_diagnostics.ipynb
- Nhiệm vụ: so sánh hiệu năng của danh mục/mã so với VNINDEX hoặc market proxy.
- Đầu ra: NAV spread, benchmark tracking diagnostics.

### 11_trade_card_generation.ipynb
- Nhiệm vụ: kết xuất thẻ giao dịch cụ thể chứa mức giá đóng cửa, stop-loss và câu nhận định chi tiết từ dữ liệu.
- Đầu ra: breakout_20d_trade_cards.parquet.

### 12_meta_ranking_optimization.ipynb
- Nhiệm vụ: hợp nhất điểm breakout setup, tích lũy stealth, momentum chéo và rủi ro thành Meta Score duy nhất.
- Đầu ra: Bảng xếp hạng tổng hợp meta_ranking.parquet.

### 13_advanced_quant_analysis.ipynb
- Nhiệm vụ: phân tích tương quan trượt (rolling correlation), rolling beta, và phát hiện phân kỳ cặp (pair divergence).
- Đầu ra: ma trận tương quan theo regime, các cặp giao dịch chênh lệch giá tiềm năng.

### 14_portfolio_construction.ipynb
- Nhiệm vụ: thiết lập danh mục mục tiêu theo các thuật toán tối ưu hóa (Equal Weight, Inverse Vol, Risk Parity, HRP).
- Đầu ra: target weights bảng phân bổ danh mục.

### 15_risk_management.ipynb
- Nhiệm vụ: áp dụng cơ chế quản trị rủi ro overlay (kiểm soát drawdown lăn, điều chỉnh hạ tỷ trọng khi vi phạm).
- Đầu ra: overlay weights bảng tỷ trọng danh mục sau quản trị rủi ro.

### 16_capital_allocation.ipynb
- Nhiệm vụ: tính toán khối lượng vị thế (position sizing) chi tiết dựa trên rủi ro tối đa và Kelly criterion.
- Đầu ra: kế hoạch phân bổ vốn chi tiết (capital allocation plan).

### 17_live_watchlist_dashboard.ipynb
- Nhiệm vụ: xây dựng giao diện dashboard động gom tất cả các bước từ sàng lọc mã sang tỷ trọng phân bổ vốn.
- Đầu ra: Live dashboard preview.

### 18_client_watchlist_report.ipynb
- Nhiệm vụ: tạo báo cáo đẹp mắt, trực quan đóng gói danh sách watchlist và giải thích logic để gửi khách hàng.
- Đầu ra: Client-ready report.
